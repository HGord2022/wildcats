```python
#simulated data
filename = "../simulation/output/round_1_point/summary_stats_r1_15k.csv"
x = pd.read_csv(filename)
```

```python
to_drop = [ # >99% correlated
  "y3_domestic_scot_lyb",
  "y3_domestic_captive_lyb",
  "diversity_domestic",
  "f3_domestic_scot_captive",
  "f3_domestic_scot_eu",
  "f3_domestic_captive_eu",
  "y3_domestic_eu_lyb",
  "divergence_domestic_eu",
  "divergence_domestic_captive",
  "divergence_scot_lyb",
  "segregating_sites_eu",
  "diversity_lyb",
  "pc2_median_lyb"
]
```

```python
x = x.drop(columns=to_drop)
np.shape(x)
```

```python
df = x.corr().abs()
s = df.unstack()
so = s.sort_values(kind="quicksort", ascending=False)
```

```python
print(so[110:220])
```

```python
import networkx as nx

threshold = 0.99

# Reset the MultiIndex to get a proper dataframe
df_long = so.reset_index()
df_long.columns = ["stat1", "stat2", "corr"]

# Build graph
G = nx.Graph()
for _, row in df_long.iterrows():
    s1, s2, r = row["stat1"], row["stat2"], row["corr"]
    if s1 != s2 and r >= threshold:
        G.add_edge(s1, s2)

# Greedy removal
to_remove = []
while G.number_of_edges() > 0:
    node = max(G.degree, key=lambda x: x[1])[0]  # node with highest degree
    to_remove.append(node)
    G.remove_node(node)

# Lists
to_remove_list = [f'"{stat}"' for stat in to_remove]
to_keep_list = [f'"{stat}"' for stat in sorted(set(df_long["stat1"]).union(df_long["stat2"]) - set(to_remove))]

print("to_remove = [")
for stat in to_remove_list:
    print(f"  {stat},")
print("]\n")

print("to_keep = [")
for stat in to_keep_list:
    print(f"  {stat},")
print("]")
```
